# 01 — EDA NOAA Gulf of Mexico
Exploration rapide du fichier CSV NOAA AccessAIS (Golfe du Mexique).

```python
import pandas as pd
import glob
from pathlib import Path

# Racine du projet (remonte d'un niveau depuis notebooks/)
PROJECT_ROOT = Path.cwd().parent

# Trouve automatiquement le fichier CSV dans le dossier
folder = PROJECT_ROOT / "data" / "raw" / "noaa_gulf"
csv_files = glob.glob(str(folder / "*.csv"))
print("Fichiers trouvés :", csv_files)
```

```python
path = csv_files[0]

# Charge le fichier
df = pd.read_csv(path)
```

```python
# 1. Dimensions
print("Nombre de lignes :", len(df))
print("Nombre de colonnes :", df.shape[1])
```

```python
# 2. Colonnes disponibles
df.columns.tolist()
```

```python
# 3. Aperçu des 5 premières lignes
df.head()
```

```python
# 4. Nombre de bateaux uniques (MMSI)
print("Nombre de MMSI uniques :", df['MMSI'].nunique())
```

```python
# 5. Types de navires présents
df['VesselType'].value_counts().head(10)
```

```python
# 6. Repérer les valeurs sentinelles (données manquantes déguisées)
print("Valeurs COG = 360.0 :", (df['COG'] == 360.0).sum())
print("Valeurs SOG = 102.3 :", (df['SOG'] == 102.3).sum())
```
